```python
import numpy as np
import matplotlib.pyplot as plt
from math import erf

x = np.linspace(-4, 4, 800)
erf_vec = np.vectorize(erf)
phi_cdf = 0.5 * (1 + erf_vec(x / np.sqrt(2)))
gelu = x * phi_cdf

relu = np.maximum(0, x)

plt.figure(figsize=(8,5))
plt.plot(x, gelu, label="GELU(x)")
plt.plot(x, relu, linestyle="--", label="ReLU(x)")
plt.axhline(0, linewidth=0.7)
plt.axvline(0, linewidth=0.7)
plt.xlabel("x")
plt.ylabel("Activation output")
plt.title("GELU forward pass")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```

```python
pdf = (1 / np.sqrt(2*np.pi)) * np.exp(-x**2 / 2)
gelu_prime = phi_cdf + x * pdf

plt.figure(figsize=(8,5))
plt.plot(x, phi_cdf, label="Φ(x): Gaussian CDF")
plt.plot(x, pdf, linestyle="--", label="φ(x): Gaussian PDF")
plt.axhline(0, linewidth=0.7)
plt.axvline(0, linewidth=0.7)
plt.xlabel("x")
plt.ylabel("Value")
plt.title("The two Gaussian functions behind GELU")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.plot(x, gelu_prime, label="GELU'(x)")
plt.axhline(0, linewidth=0.7)
plt.axhline(1, linestyle="--", linewidth=0.7)
plt.axvline(0, linewidth=0.7)
plt.xlabel("x")
plt.ylabel("Local gradient")
plt.title("GELU backward derivative")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```
